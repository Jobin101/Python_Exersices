```python
# Creating sets
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Basic modification
set_a.add(7)          #add()
print(set_a)
        # set_a is now {1, 2, 3, 4, 7}
set_a.discard(2)         #discard()
print(set_a)

print(set_a & set_b)
```

```python
# Defining our sets.   #Boolean
group_a = {1, 2, 3}
group_b = {1, 2, 3, 4, 5}
group_c = {6, 7, 8}

# 1. isdisjoint() - Checks for ZERO common elements

# True because group_a and group_c share no elements
print(group_a.isdisjoint(group_c))  # Output: True
# False because group_a and group_b share {1, 2, 3}
print(group_a.isdisjoint(group_b))  # Output: False

# 2. issubset() - Checks if all elements exist in the target set

# True because every element in group_a ({1, 2, 3}) is also in group_b
print(group_a.issubset(group_c))    # Output: false
# You can also use the '<=' operator
print(group_a <= group_c)           # Output: false

# 3. issuperset() - Checks if the set contains all elements of the target set

# True because group_b contains everything that is in group_a
print(group_b.issuperset(group_a))  # Output: True
# You can also use the '>=' operator
print(group_b >= group_a)           # Output: True
```

```python
# Python Set Methods and Functions

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Set A:", a)
print("Set B:", b)

# 1. add()
a.add(7)
print("add():", a)

# 2. update()
a.update([8, 9])
print("update():", a)

# 3. remove()
a.remove(9)
print("remove():", a)

# 4. discard()
a.discard(10)
print("discard():", a)

# 5. pop()
x = a.pop()
print("pop():", x)
print("Set after pop():", a)

# 6. copy()
c = a.copy()
print("copy():", c)

# 7. clear()
temp = c.copy()
temp.clear()
print("clear():", temp)



# 18. Built-in functions
s = {10, 20, 30, 40}
print("len():", len(s))
print("min():", min(s))
print("max():", max(s))
print("sum():", sum(s))
print("sorted():", sorted(s))
print("30 in s:", 30 in s)
print("50 not in s:", 50 not in s)
```

```python
# Python Dictionary Methods and Functions

d = {"name": "Arun", "age": 21, "city": "Kerala"}

print("Dictionary:", d)

# 1. get()
print("get():", d.get("name"))

# 2. keys()
print("keys():", d.keys())

# 3. values()
print("values():", d.values())

# 4. items()
print("items():", d.items())

# 5. update()
d.update({"age": 22})
print("update():", d)

# 6. pop()
x = d.pop("age")
print("pop():", x)
print("After pop():", d)

# 7. popitem()
x = d.copy()
print("popitem():", x.popitem())
print("After popitem():", x)

# 8. setdefault()
d.setdefault("course", "Python")
print("setdefault():", d)

# 9. copy()
c = d.copy()
print("copy():", c)

# 10. clear()
c.clear()
print("clear():", c)

# 11. fromkeys()
keys = ["a", "b", "c"]
new_dict = dict.fromkeys(keys, 0)
print("fromkeys():", new_dict)

# 12. len()
print("len():", len(d))

# 13. type()
print("type():", type(d))

# 14. sorted()
print("sorted():", sorted(d))

# 15. Membership operators
print("'name' in d:", "name" in d)
print("'salary' not in d:", "salary" not in d)

# 16. del
temp = d.copy()
del temp["city"]
print("After del:", temp)
```

```python
# Recurssion: a function calling itself

#wap to find a factorial of a no. using recurssion

n=int(input("enter a no."))

def fac(n):
  if n==1:
    return 1
  else:
    return n*fac(n-1)
fac(n)

```

```python
#wap to find  sum of first n natural nos. using reccuirssion

n=int(input("enter the no."))

def sum(n):
  if n==1:
    return 1
  else:
    return n+sum(n-1)
sum(n)

```

```python
#create a class for rectngles and create methods to display its area and perimeter

class rectangle():
  def get_data(sel):
    sel.length=int(input("Enter the length:"))
    sel.breadth=int(input("enter the bteadth:"))

  def display_data(sel):
    print(sel.length)
    print(sel.breadth)

  def area (sel):
    print(sel.length*sel.breadth)

  def perimeter(sel):
    print(2*(sel.length+sel.breadth))


r=rectangle()
r.get_data()
r.display_data()
r.area()
r.perimeter()


```

```python
# constructor __init__ :  Used to assign values to data members, automatically cals when object is created.

class rectangle():
  def __init__(sel,l,b):
    sel.length=l
    sel.breadth=b

  def display_data(sel):
    print(sel.length)
    print(sel.breadth)

  def area (sel):
    print(sel.length*sel.breadth)

  def perimeter(sel):
    print(2*(sel.length+sel.breadth))


r=rectangle(12,16)

r.display_data()
r.area()
r.perimeter()

```
